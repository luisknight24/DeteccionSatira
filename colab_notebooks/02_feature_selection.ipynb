# Fase 2: Selección de Características

Este notebook reduce la dimensionalidad del dataset original de características (163 columnas) a las 23 columnas finales (19 características seleccionadas + 4 metadatos), eliminando las columnas fraseológicas individuales que generan alta dispersión y posible sobreajuste, para dejar las características NLP más significativas.

```python
# 1. Conexión con Google Drive
from google.colab import drive
drive.mount('/content/drive')
```

```python
# 2. Carga de librerías
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# 3. Cargar el dataset extraído de Fase 1
PATH = "/content/drive/MyDrive/Titulacion/DatasetsFinales/"
df_train = pd.read_json(PATH + "df_train2_feat.jsonl", orient='records', lines=True)
print(f"Dimensiones iniciales: {df_train.shape}")
```

```python
# 4. Definición de características seleccionadas
metadata_cols = ['id', 'label', 'transcription', 'transcription_processed']

selected_features = [
    'MeanWordLen',
    'LexicalDiversity',
    'MeanSentenceLen',
    'StdevSentenceLen',
    'DocumentLen',
    'WordsPerText',
    'SentencesPerText',
    'num_words',
    'num_chars',
    'irony_score',
    'prop_NOUN',
    'prop_VERB',
    'prop_ADJ',
    'rhetorical_questions',
    'avg_depth',
    'Flesch Score',
    'Lexical Entropy',
    'Syntactic Repetition',
    'Unusual Word Frequency'
]

final_cols = metadata_cols + selected_features
```

### Análisis Exploratorio y Filtrado

```python
# Filtrar dataset
df_selected = df_train[final_cols]
print(f"Dimensiones finales tras selección: {df_selected.shape}")
```

```python
# Visualizar matriz de correlación de las características seleccionadas
plt.figure(figsize=(14, 12))
corr = df_selected[selected_features].corr()
sns.heatmap(corr, annot=True, fmt=".2f", cmap='coolwarm', cbar=True, square=True)
plt.title('Matriz de Correlación de Características Seleccionadas')
plt.show()
```

```python
# Guardar dataset final
OUTPUT_PATH = PATH + "df_train2_featselect2.jsonl"
df_selected.to_json(OUTPUT_PATH, orient='records', lines=True)
print(f"Dataset seleccionado guardado en: {OUTPUT_PATH}")
```
